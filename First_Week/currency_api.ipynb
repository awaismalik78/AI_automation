```python
import requests

url = "https://open.er-api.com/v6/latest/USD"

response = requests.get(url)

print(response.status_code)
print(response.text)
```

```python
data = response.json()

print(type(data))
```

```python
print("Base:", data["base_code"])

print("PKR rate:", data["rates"]["PKR"])
print("EUR rate:", data["rates"]["EUR"])
```

```python
usd_amount = 10

pkr_rate = data["rates"]["PKR"]

pkr_amount = usd_amount * pkr_rate

print("USD:", usd_amount)
print("PKR:", pkr_amount)
```

```python
import requests

url = "https://open.er-api.com/v6/latest/USD"

data = requests.get(url).json()

amount = float(input("Enter USD amount: "))

pkr_rate = data["rates"]["PKR"]

print("PKR:", amount * pkr_rate)
```

```python
import requests

try:
    url = "https://open.er-api.com/v6/latest/USD"
    response = requests.get(url, timeout=5)

    if response.status_code != 200:
        print("API Error:", response.status_code)
        exit()

    data = response.json()

    if "rates" not in data:
        print("Invalid response format")
        exit()

    amount = float(input("Enter USD: "))

    print("PKR:", amount * data["rates"]["PKR"])

except requests.exceptions.Timeout:
    print("Request timed out")

except requests.exceptions.ConnectionError:
    print("No internet connection")

except ValueError:
    print("Invalid input number")

except Exception as e:
    print("Unexpected error:", e)
```
